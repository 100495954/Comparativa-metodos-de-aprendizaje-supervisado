```python
import pandas as pd
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Cargar modelo final
modelo_final = joblib.load("modelo_final.pkl")

# Cargar datos de competición
# Asegúrate de que el archivo tenga las mismas columnas (excepto la etiqueta)
datos_competencia = pd.read_csv("attrition_competition_08.csv.gz")

# Realizar predicciones
predicciones = modelo_final.predict(datos_competencia)

# Guardar predicciones en un CSV
# Suponemos que se predice una sola etiqueta por fila
pd.DataFrame(predicciones, columns=["Prediccion"]).to_csv("predicciones.csv", index=False)

print("Predicciones guardadas en 'predicciones.csv'")

# Contar cuántas veces aparece cada clase
valores, cuentas = np.unique(predicciones, return_counts=True)

# Crear gráfico
plt.figure(figsize=(6, 4))
sns.barplot(x=valores, y=cuentas, palette="viridis")

plt.xticks([0, 1], ["No", "Sí"])
plt.ylabel("Cantidad")
plt.title("Distribución de Predicciones (No vs Sí)")
plt.tight_layout()
plt.show()
```
